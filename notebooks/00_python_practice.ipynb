## Dictionaries and DataFrames

A dictionary stores information using key-value pairs. In this practice example, each dictionary represents one fictional donor.

Multiple donor dictionaries can be combined into a list and converted into a pandas DataFrame.

A DataFrame organizes information into rows and columns and will be useful for organizing donor metadata and analysis results in the SEA-AD project.

**Note:** All donor information and values in this section are fictional and are used only for Python practice.

```python
len(df)
len(df)
df["donor_id"].nunique()
df["pathology_group"].value_counts()
df.isna()
df.isna().sum()
```

```python
practice_df = df.copy()

practice_df.loc[1, "age"] = None

practice_df
practice_df.isna().sum()
df.groupby("pathology_group")["age"].mean()
df.groupby("pathology_group")["interferon_score"].median()
df.groupby("pathology_group")["interferon_score"].agg(["count", "mean", "median"])
```

```python
import matplotlib.pyplot as plt
plt.scatter(
    df["pathology_group"],
    df["interferon_score"]
)

plt.xlabel("Pathology Group")
plt.ylabel("Interferon Score")
plt.title("Interferon Scores by Pathology Group")

plt.show()
group_means = df.groupby("pathology_group")["interferon_score"].mean()

print(group_means)
group_means.plot(kind="bar")

plt.xlabel("Pathology Group")
plt.ylabel("Mean Interferon Score")
plt.title("Mean Interferon Score by Pathology Group")

plt.show()
```
